```python
# Load the original image
image = cv2.imread('lilies (1).jpg')
# Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
# Save the grayscale image with a new filename
cv2.imwrite('grayscale_output.jpg', gray_image)
print("Grayscale image saved as 'grayscale_output.jpg'")

# Download the file to your computer
files.download('grayscale_output.jpg')
```

```python
import cv2
import os

# Check if gray_image exists in memory; if not, try to load it from the saved file
if 'gray_image' not in globals():
    if os.path.exists('grayscale_output.jpg'):
        gray_image = cv2.imread('grayscale_output.jpg', cv2.IMREAD_GRAYSCALE)
    else:
        raise FileNotFoundError("The file 'grayscale_output.jpg' could not be found. Please re-run the previous cell to regenerate the image.")

# Get the dimensions of the grayscale image
height, width = gray_image.shape

# Initialize an empty 2D list to store brightness values
brightness_array = []

# Loop over every pixel
for r in range(height):
    row_values = []
    for c in range(width):
        # Get pixel brightness at row r and column c
        pixel_value = int(gray_image[r, c])
        row_values.append(pixel_value)
    brightness_array.append(row_values)

print(f"Successfully created a 2D array of size: {len(brightness_array)}x{len(brightness_array[0])}")
```

```python
import cv2
from google.colab import files
import os

# 1. Ask the user to upload a new colored image
print("Please upload a colored image:")
uploaded_color = files.upload()

# Get the filename of the uploaded image
color_filename = list(uploaded_color.keys())[0]

# 2. Load the image in color (default BGR format in OpenCV)
color_image = cv2.imread(color_filename)

# Get the dimensions of the colored image
color_height, color_width, channels = color_image.shape

# 3. Initialize an empty 2D list to store color values
color_pixel_array = []

# Loop over every pixel to store its BGR color values
for r in range(color_height):
    row_colors = []
    for c in range(color_width):
        # Get the B, G, R components at row r and column c
        bgr_pixel = color_image[r, c]
        # Convert numpy array values to standard Python integers
        pixel_color = [int(bgr_pixel[0]), int(bgr_pixel[1]), int(bgr_pixel[2])]
        row_colors.append(pixel_color)
    color_pixel_array.append(row_colors)

print(f"\nSuccessfully created a 2D color pixel array of size: {len(color_pixel_array)}x{len(color_pixel_array[0])}")
print("Each cell in this array contains [Blue, Green, Red] intensity values.")
```

```python
import json
from google.colab import files

# 1. Save the grayscale array to greyarray.txt
with open('greyarray.txt', 'w') as f:
    # Using json.dump to format the list structure cleanly in the file
    json.dump(brightness_array, f)
print("Saved grayscale array to 'greyarray.txt'")

# 2. Save the color array to colorarray.txt
with open('colorarray.txt', 'w') as f:
    json.dump(color_pixel_array, f)
print("Saved colored array to 'colorarray.txt'")

# 3. Download both files to your computer
files.download('greyarray.txt')
files.download('colorarray.txt')
```
